```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

df = sns.load_dataset("taxis")
df.head()
```

```python
df["pickup"] = pd.to_datetime(df["pickup"])
df["dropoff"] = pd.to_datetime(df["dropoff"])
```

```python
df.dtypes
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numeric_columns)
```

```python
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
categorical_columns = df.select_dtypes(include="object").columns

print("Categorical Columns:")
print(categorical_columns)
```

```python
for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df = df.dropna(subset=["pickup", "fare", "distance"])
```

```python
df = df.dropna(subset=["pickup", "fare", "distance"])
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["pickup"], df["fare"])

plt.xlabel("Pickup Time")
plt.ylabel("Fare")
plt.title("Taxi Fare Over Time")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

Bar chart

```python
borough_fare = df.groupby("pickup_borough")["fare"].sum()

borough_fare
```

```python
plt.figure(figsize=(10, 6))

borough_fare.plot(kind="bar")

plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare")
plt.title("Total Fare by Pickup Borough")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

Pie chart

```python
payment_counts = df["payment"].value_counts()

payment_counts
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Trips by Payment Method")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["distance"], bins=30)

plt.xlabel("Distance")
plt.ylabel("Number of Trips")
plt.title("Distribution of Taxi Trip Distance")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

df.boxplot(
    column="tip",
    by="pickup_borough"
)

plt.xlabel("Pickup Borough")
plt.ylabel("Tip")
plt.title("Tip Distribution by Pickup Borough")

plt.suptitle("")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

SEABORN VISUALIZATIONS

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="pickup_borough"
)

plt.xlabel("Pickup Borough")
plt.ylabel("Number of Trips")
plt.title("Number of Taxi Trips by Pickup Borough")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="distance",
    y="fare",
    hue="pickup_borough")

plt.xlabel("Distance")
plt.ylabel("Fare")
plt.title("Relationship Between Distance and Fare")

plt.tight_layout()

plt.show()
```

```python
correlation = df[
    ["distance", "fare", "tip", "tolls", "total"]
].corr()

correlation
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()

plt.show()
```
